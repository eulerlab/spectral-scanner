```python
import os
import pyboard
```

```python
## find out the port that connects to ESP 32
result = os.popen('ls -l /dev/ttyUSB*').read()
result = result.strip('\n').split(' ')
port = result[-1]
print(port)
```

```python
import serial

# Replace '/dev/ttyUSB0' with your ESP32 port (Windows: 'COMx', Mac/Linux: '/dev/ttyUSBx')
ser = serial.Serial('/dev/ttyUSB2', 115200, timeout=1)
```

```python
pb = None
try:
    pb = pyboard.Pyboard(port)
    pb.enter_raw_repl()  
except pyboard.PyboardError as e:
    print(e) 
```
